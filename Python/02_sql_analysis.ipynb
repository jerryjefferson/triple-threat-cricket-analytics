```python
import sqlite3
import pandas as pd
```

```python
player_performance = pd.read_csv(
    "../Cleaned Data/player_performance.csv"
)

team_summary = pd.read_csv(
    "../Cleaned Data/team_summary.csv"
)
```

```python
conn = sqlite3.connect(
    "../SQL/triple_threat.db"
)
```

```python
player_performance.to_sql(
    "player_performance",
    conn,
    if_exists="replace",
    index=False
)

team_summary.to_sql(
    "team_summary",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT *
FROM player_performance
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT "Player Name", "Team Name", Batting_Runs, Batting
FROM player_performance
WHERE Batting_Runs >= 50
ORDER BY Batting_Runs DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT "Player Name", "Team Name", Batting_Runs
FROM player_performance
ORDER BY Batting_Runs DESC
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT  "Team Name", SUM(Batting_Runs) AS Total_runs
FROM player_performance
GROUP BY "Team Name"
ORDER BY Total_Runs DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT  "Team Name", SUM(Wickets) AS Total_wickets, SUM(Bowling_Runs) AS Runs_Conceded,
    ROUND(SUM(Bowling_Runs) * 6.0 / SUM(Overs_Balls), 2) AS Economy
FROM player_performance
GROUP BY "Team Name"
ORDER BY Total_wickets DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT COUNT(*)
FROM pragma_table_info('player_performance');
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT  "Player Name", "Team Name", ROUND((Batting_Runs/Balls)*100,2) AS strike_rate
FROM player_performance
ORDER BY strike_rate DESC
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT  "Player Name", "Team Name", ROUND((Batting_Runs/Balls)*100,2) AS strike_rate
FROM player_performance
ORDER BY strike_rate DESC
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT  "Player Name", "Team Name", Batting_strike_rate,
CASE WHEN Batting_strike_rate >= 200 THEN "Excellent"
     WHEN Batting_strike_rate >= 150 THEN "Good"
ELSE "Average" END AS Batting_Performance
FROM player_performance
WHERE Batting_runs >= 30
;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    "Player Name",
    "Team Name",
    Wickets,
    Bowling_Runs,
    ROUND(Bowling_Runs * 6.0 / Overs_Balls, 2) AS Economy
FROM player_performance
WHERE Wickets > 0
ORDER BY Wickets DESC, Economy ASC
;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT  "Player Name", "Team Name", Batting_runs,
DENSE_RANK() OVER(ORDER BY Batting_runs DESC) AS rnk
FROM player_performance
WHERE Batting_runs IS NOT NULL

;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT  "Team Name", SUM(Batting_runs) AS total_runs,
DENSE_RANK() OVER(ORDER BY SUM(Batting_runs) DESC) AS team_rnk
FROM player_performance
Group BY "Team Name"
ORDER BY team_rnk
;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    "Player Name",
    "Team Name",
    Batting_Runs,
    Wickets
FROM player_performance
WHERE Batting_Runs >= 30
  AND Wickets >= 1
ORDER BY Batting_Runs DESC, Wickets DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
with econ AS(
SELECT "Player Name",
    "Team Name",
    Wickets,
    Batting_runs,
    Fielding_Dismissals,
    Bowling_Runs,
    ROUND(Bowling_Runs * 6.0 / Overs_Balls, 2) AS Economy
FROM player_performance
)
SELECT
    "Player Name",
    "Team Name",
    Batting_Runs,
    Wickets,
    Fielding_Dismissals,
    Economy
FROM econ
WHERE Batting_Runs >= 20
  AND Wickets >= 1
  AND Fielding_Dismissals >= 1
  AND Economy < 15
ORDER BY Batting_Runs DESC, Wickets DESC, Fielding_Dismissals DESC, Economy ASC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    "Player Name",
    "Team Name",
    Batting_Score,
    Bowling_Score,
    Fielding_Score,
    Overall_Score,
    DENSE_RANK() OVER(ORDER BY Overall_Score DESC) AS Overall_Rank
FROM player_performance
WHERE Overall_Score IS NOT NULL
ORDER BY Overall_Rank;
"""

pd.read_sql_query(query, conn)
```
